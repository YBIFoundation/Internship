In this project, the Titanic dataset is used for data preprocessing and analysis.

It is a well-known dataset widely used in AI and Data Science for learning data preprocessing and exploratory data analysis.

```python
# Load Pandas library
import pandas as pd

# Read Titanic dataset
df = pd.read_csv("https://raw.githubusercontent.com/YBIFoundation/Dataset/main/Titanic.csv")
df.head()   # Display first 5 rows
```

```python
# Check missing values in dataset
df.isnull().sum()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Histogram for age
plt.figure(figsize=(8,5))
sns.histplot(df['age'], bins=30, kde=True, color='skyblue')
plt.xlabel("age")
plt.ylabel("Count")
plt.title("age Distribution")
plt.show()
```

```python
# 1. Fill missing values in 'age' with the average (mean) age
df['age'] = df['age'].fillna(df['age'].mean())

# 2. Fill the 1 missing value in 'fare' with the median value
df['fare'] = df['fare'].fillna(df['fare'].median())

# 3. Drop the 'cabin' column because it has too many missing values (1014)
df.drop(columns=['cabin'], inplace=True)

print("Missing values have been handled successfully!")
```

```python
# Check the missing value count again
print(df.isnull().sum())
```
